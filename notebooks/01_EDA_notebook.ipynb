```python
import pandas as pd
import plotly.express as px
pd.options.display.max_colwidth = 200
# Load feature-engineered CSV (created by the pipeline)
df = pd.read_csv('../data/feature_reviews.csv', parse_dates=['time'], low_memory=False)
df.head()
```

```python
# Quick shape and schema
print('Rows, cols:', df.shape)
df.info()
df.describe(include='all')
```

```python
# Missing values and duplicates
print('Missing per column:
', df.isna().sum())
print('Number of exact duplicate rows:', df.duplicated().sum())
# Optionally drop duplicates in-place for analysis
df = df.drop_duplicates()
print('After dropping duplicates:', df.shape)
```

```python
# Rating distribution
fig = px.histogram(df, x='rating', nbins=5, title='Rating distribution')
fig.show()
```

```python
# Review length analysis (words & chars)
print(df['word_count'].describe())
fig = px.histogram(df, x='word_count', nbins=50, title='Review word count distribution')
fig.show()
# Compare word count by rating
fig = px.box(df, x='rating', y='word_count', points='all', title='Word count by rating')
fig.show()
```

```python
# Lexical diversity & repetition by rating
lex_rep = df.groupby('rating')[['lexical_diversity','repetition_ratio']].mean().round(3)
lex_rep
```

```python
# POS ratios (adj/noun/verb) by rating
pos_metrics = df.groupby('rating')[['adj_ratio','noun_ratio','verb_ratio']].mean().round(3)
pos_metrics
```

```python
# Temporal coverage and review volume over time
print('Time range:', df['time'].min(), df['time'].max())
ts = df.set_index('time').resample('M').size().reset_index(name='count')
fig = px.line(ts, x='time', y='count', title='Monthly review volume')
fig.show()
```
